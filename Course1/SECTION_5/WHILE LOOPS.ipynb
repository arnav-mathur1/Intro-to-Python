```python
p = 1; o = 1
while p < 15 and o < 15:
    print("%i / %f = %i" % (p, o, (p/o)))
    p += 1; o += 1
```

```python
num = 5; cake = 10
while num < 12:
    while cake < 17:
        print("{0} + {1} + {0} = {2}".format(num, cake, num, num+cake))
        cake += 1; num += 1
```
