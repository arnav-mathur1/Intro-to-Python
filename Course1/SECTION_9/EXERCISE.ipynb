```python
d = open("rain.txt", "w") 
d.write("Hi") 
d.write("\nfood is hot"*2) 
d.write("\nsunny day"*4) 
d.close() 

e = open("cloud.txt", "w") 
e.write("Hi") 
e.write("\nfood is cold"*2) 
e.write("\nsunny day"*4) 
e.close()
```

```python
a = open("rain.txt", "r")
b = open("cloud.txt", "r")
day = a.readlines() 
night = b.readlines() 
for i in range(6):
    if day[i] == night[i]:
        print("True")
    else:
        print("False")
a.close()
b.close()
```
