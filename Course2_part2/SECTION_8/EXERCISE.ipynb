```python
path = "C:/Users/Arnav/Desktop/" 
def Where_is_my_pet():    
    import os    
    txt_folder = os.listdir(path + "ANIMAL_SHELTER")    
    x = 0    
    while x < len(txt_folder):
        pet  = input("What animal is in this house? ").lower()        
        with open(path + "ANIMAL_SHELTER/House_{}.txt".format(x), "r") as f:         
            reading = f.readlines()            
            search = "".join(reading).replace("\n", " ").split()            
            if pet in search:               
                print("Found my {} in  House {}".format(pet, x))                
                print("--" * 15)            
            elif pet == "quit":                
                print("I give up")                
                break 
            else:                
                print("Lets look in the next House")               
                print("*" * 15)                
                continue
        x += 1    
    print("End the search")

Where_is_my_pet()
```

```python
def Professor(*args, path = "C:/Users/Arnav/Desktop/", folder_name):    
    # PATH.EXISTS PREVENTS FILE EXISTS ERROR    
    if not os.path.exists(path + folder_name):          
        os.mkdir(path + folder_name)    
        
    students = [i for i in args]
    
    for i in range(len(students)):        
        percent = input("Student: {} Grade = ".format(students[i]))       
        mark = eval(percent)        
        
        if mark >= 70:            
            grade = "first class"        
        elif mark >= 60 and mark < 70:            
            grade = "second class"        
        elif mark >= 50 and mark < 60:            
            grade = "lower second class"        
        else:            
            grade = "third class"        
            
        with open(path +  folder_name + "\Award_for_{}.txt".format(students [i]), "w") as f:     
             f.write(f""""Dear {students[i]}, \t
             You've achieved {percent}%, so you've been awarded a {grade}!""")
```
