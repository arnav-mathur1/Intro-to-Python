```python
import os, pickle
path = "C:/Users/Arnav/Desktop/"
pick = "PICKLED_FOLDER"

a = [10, 20, 30]
b = ("cat", "dog", "cow")
c = {"x": a, "y": b}
data = [a, b, c]
```

```python
os.mkdir(path + pick)
```

```python
os.listdir(path)
```

```python
loc = input("Create a new folder: ").upper()
if loc not in os.listdir(path):
    os.mkdir(path + loc)
    print("New folder created!")
else:
    print()
    print("{} already exists!".format(loc))
```

```python
for i in range(3):
    os.removedirs(path + "OS_{}".format(i))
```

```python
for i in range(len(data)):
    
    with open(path + pick + f"/var_{i}.pickle", "wb") as w:
        pickle.dump(data[i], w)
    
    with open(path + pick + f"/finance_{i}.txt", "w") as f:
        f.write("\n" + str(data[i]))
```

```python
res = []
folder_name = input("Enter name of folder: ").upper()
file_name = input("Enter name of file format: ")

for f in os.listdir(path + "{}".format(folder_name)):
    if f.endswith(".{}".format(file_name)):
        res.append(f)
res
```
