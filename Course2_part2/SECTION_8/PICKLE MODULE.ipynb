```python
import pickle, numpy as np
path = "C:/Users/Arnav/Desktop/PICKLE_FOLDER/"
```

```python
num = np.array([10, 20, 30, 40])

keys = ["x", "y", "z"]

d = {"a": num*2, "b": num*3, "c": num*4}

t = tuple(list(range(0, 110, 10)))

bt = pickle.dumps(num)
bt

data = [num, keys, d, t]
data
```

```python
for i in range(len(data)):
    
    with open(path + "pick_me_{}.pickle".format(i), "wb") as w:
        pickle.dump(data[i], w)
```

```python
unpickled = []
for i in range(len(data)):
    
    with open(path + "pick_me_{}.pickle".format(i), "rb") as r:
        res = pickle.load(r)
        unpickled.append(res)
unpickled
```
