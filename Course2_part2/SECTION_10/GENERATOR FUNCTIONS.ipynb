```python
def func1(n):
    case = []
    for num in range(n):
        res = num**2
        case.append(res)
    return case
# func1(5)

f = iter(fun1(5))
f
```

```python
list(f)
```

```python
def gen1(n):
    for num in range(n):
        yield num**2
        
g = gen1(5)
g
```

```python
type(next)
next(g)
```

```python
list(g)
```

```python
def GoT(val, call):
    
    num1 = val - 8
    for x in range(num1):
        word1 = ["Game of Thrones" for i in range(call)]
        yield word1
        
        num2 = val - 7
        for s in range(num2):
            word2 = ["King of the North!" for k in range(call)]
            yield word2
```

```python
game = GoT(10, 2)
game
```

```python
list(game)
```

```python
next(game)
```

```python
def multiple(n):
    yield n
    yield n/10
    yield n + 10
    yield n * 10 
    yield n**2 
m = multiple(4)
m   
```

```python
for i in list(m):
    print(i)
```
