```python
def gen_func():
    for i in range(3, 10):
        yield i**2
        
g = gen_func()
print(g)
list(g)
```

```python
gen = (i**2 for i in range(3, 10))
print(gen)
list(gen)
```

```python
list_comp = [i**2 for i in range(3, 10)]
list_comp
```

```python
gen_list = iter(list_comp)
print(gen_list)
```

```python
def fun3(n):
    for i in range(n):
        if i < 2:
            print("COOL")           
        elif i >= 2 and i < 4:
            print("AWESOME!")
        else:
            print("SUPER!")
fun3(5)
```

```python
gen_cf = ("COOL" if i < 2 else "AWESOME!" if i >= 2 and i < 4 else "SUPER!" for i in range(5))
print(gen_cf)
for i in gen_cf:
    print(i)
```

```python
def gen2():
    for i in range(10, 12):
        yield i*20
        yield i + 100
        
g = gen2()
print(g)
list(g)
```

```python
gen_exp = ((i*20, i+100) for i in range(10, 12))
print(gen_exp)
list(gen_exp)
```

```python
gen_exp2 = (x for i in range(10, 12) for x in (i*20, i+100))
print(gen_exp2)
list(gen_exp2)
```

```python
for i in range(10, 12):
    for x in (i*20, i+100):
        print(x)
```

```python
gen_exp3 = ((v*20, v+100, v*4) for v in range(1, 4))
print(gen_exp3)
```
