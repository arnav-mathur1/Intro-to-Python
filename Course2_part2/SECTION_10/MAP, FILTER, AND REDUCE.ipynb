```python
def square(n):
    return n**2
numbers = list(range(2, 10))
numbers
```

```python
list(map(square, numbers))
```

```python
list(map((lambda x : x**2), numbers))
```

```python
case = []; num = 10, 20, 30; power = 2, 3, 4
for i in range(3):
    s = pow(num[i], power[i])
    case.append(s)
    
case
```

```python
animals = "cat", "dog", "bird", "horse", "elephant", "dolphin"
list(map(len, animals))
```

```python
case2 = []
for i in range(len(animals)):
    s = len(animals[i])
    case2.append(s)
    
case2
```

```python
temp = list(range(-10, 10))
temp
```

```python
list(filter((lambda x : x > -5 and x < 5), temp))
```

```python
case3 = []
for i in temp:
    if i > -5 and i < 5:
        res = i
        case3.append(res)
        
case3
```
