```python
numbers = [10, 23, 30, 33, 40, 43]
for i in numbers:
    if i % 2 == 0:
        print(i**2)
    else:
        print("ODD")
```

```python
numbers = [10, 23, 30, 33, 40, 43]
g = lambda x : x**2 if x % 2 == 0 else "ODD"
g_map = list(map(g, numbers)) 

for i in g_map:
    print(i)
```

```python
def gen_list(num = 10, var = 4):
    yield num
    yield var
    yield num + 100
    yield var + 100
    yield num*var
    
n = gen_list()
list(n)
```

```python
lang = ["Python", "Javascript", "Java", "C++", "Go" "R", "VBA"]
gen = (lang[i].upper() for i in range(len(lang)) if len(lang[i]) >= 4) # EXPRESSION
list(gen)
```

```python
def gen_func(): # FUNCTION
    for i in range(len(lang)):
        if len(lang[i]) >= 4:
            res = lang[i].upper()            
            yield res
            
g = gen_func()
list(g)
```
