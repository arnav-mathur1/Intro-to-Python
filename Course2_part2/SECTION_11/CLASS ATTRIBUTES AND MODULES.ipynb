```python
class Delivery:
    service_charge = 1.15
    
    prices = [1.0, 0.85, 0.7, 0.50]
    
    def __init__(self):
        
        import random as rd
        
        self.rd = rd
        
        self.rd.shuffle(Delivery.prices)
        val = Delivery.prices[0]
       
        self.menu = {1: ["coconut curry", round(7.50*val, 2)], 
                     2: ["spicy tofu", round(5.25*val, 2)], 
                     3: ["veg noodles", round(8.75*val, 2)]} 
        self.order = []
        
    def Choose_in_menu(self):
        for i in range(1, len(self.menu) + 1):
            print(i, self.menu[i][0] + " $" + str(self.menu[i][1]))
            
        print("Please enter the dish numbers you'd like")
        
        while True:
            self.food = input("")
            try:
                num = abs(int(eval(self.food)))
                
            except:
                print("Please enter a valid number")
                continue
                
            if self.food == "" or num == 0 and num < 4:
                print("Thank you. Your order is complete")
                break               
            elif num >= 4:
                print("Please enter a number between 1 and 3")
            else:
                self.order.append(self.menu[num][1])
                
        print(self.order)
        
    def Pay(self):
        total = sum(self.order) * Delivery.service_charge
        print("That will be total of $" + str(round(total, 2)))
```

```python
d = Delivery()
d.Choose_in_menu()
```

```python
d.Pay()
```
