```python
class A:
    def mode(self):
        print("I'm inside class A")

class B(A):
    def mode(self):
        print("I'm inside class B")
        super(B, self).mode()
        
class C(B):
    def mode(self):
        print("I'm inside class C")
        super(C, self).mode()
       
class D(C):
    def mode(self):
        print("I'm inside class D")
        super(D, self).mode()
```

```python
class Animal_Kingdom:
    def __init__(self, species, action):
        self.species = species
        self.action = action
        
    def Animal(self):
        print("I am a {} and I can {}".format(self.species, self.action))
        
class Mammal(Animal_Kingdom):
    def __init__(self, species, action, name, age):
        super().__init__(species, action)
        self.age = age
        self.name = name
        
    def Info(self):
        print("My name is {} and I'm {}".format(self.name, self.age))
        
class Pets(Mammal):
    def __init__(self, species, action, name, age, favorite):
        super().__init__(species, action, name, age)
        self.favorite = favorite
    
    def Meal(self):
        print("Favorite food is {}".format(self.favorite))
        
class Furry_baby(Pets):
    def __init__(self, species, action, name, age, favorite, noise):
        super().__init__(species, action, name, age, favorite)
        self.noise = noise
    
    def Talk(self):
        print("The {} likes to {}".format(Furry_baby.__name__, self.noise))
```

```python
cat = Furry_baby("feline", "jump", "gracie", 5, "dreamies", "meow")
# dog = Pets("canine", "run", "Tommy", 5, "crunchies")
```

```python
cat.Animal()
cat.Info()
cat.Meal()
cat.Talk()
```
