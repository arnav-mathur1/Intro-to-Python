```python
class Delivery:
    service_charge = 1.15  
    extras = "sweetcorn soup"
#     prices = [1.0, 0.85, 0.7, 0.50]
    
    def __init__(self):
        
        self.menu = {1: ["coconut curry", 7.50], 
                     2: ["spicy tofu",  5.25], 
                     3: ["veg noodles", 8.75]} 
        self.order = []
        
    def Choose_in_menu(self):
        for i in range(1, len(self.menu) + 1):
            print(i, self.menu[i][0] + " $" + str(self.menu[i][1]))
            
        print("Please enter the dish numbers you'd like")
        
        while True:
            self.food = input("")
            try:
                num = abs(int(eval(self.food)))
                
            except:
                print("Please enter a valid number")
                continue
                
            if self.food == "" or num == 0 and num < 4:
                print("Thank you. Your order is complete")
                break               
            elif num >= 4:
                print("Please enter a number between 1 and 3")
            else:
                self.order.append(self.menu[num][1])
                        
    def Pay(self):
        total = sum(self.order) * Delivery.service_charge
        print("That will be total of $" + str(round(total, 2)))
        if total > 30:
            print(f"We've also included a free {Delivery.extras}")           
        else:
            None
            
    def Decision_Time(self, func):
        print("How long it takes for the customer to choose?")
        def Customer_Time(*args, **kwargs):
            import time
            start = time.time()
            dec1 = func()
            end = time.time()
            
            duration = round(((end-start)*1000)/1000,2)
            print("The time was {} seconds".format(duration))
            
            return dec1
        return Customer_Time
```

```python
d = Delivery()
dec1 = d.Decision_Time(d.Choose_in_menu)
```
