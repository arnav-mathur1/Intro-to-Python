```python
class Parent:
    """This is a parent, super, or base class"""
    A = 10
    B = 20
    C = 40
    
    def __init__(self, move, age):
        self.move = move
        self.age = age
        
    def action(self):
        print("My daily activity is {}".format(self.move))
        
    def life(self):
        print("My age is " + str(self.age))
        
class Boy(Parent):
    """This is the child, sub, or derived class"""
    
    C = 550
    D = 77
    
    def __init__(self, move, age, comicshow):
        super().__init__(move, age)
        self.comicshow = comicshow
        
    def marvel(self):
        print("I like watching {}".format(self.comicshow.upper()))
        
class Girl(Parent):
    
    def __init__(self, move, age, fashion, pose):
        super().__init__(move, age)
        self.fashion = fashion
        self.pose = pose
        
    def action(self):
        super(Girl, self).action()
        print("....And I can do the {}".format(self.pose))
    
    def Fashion(self):
        print(f"Like to wear {self.fashion}")
        
    def young_life(self):
        super(Girl, self).life()
        
        if self.age <= 16:
            print("I go to high school")
        elif self.age > 16 and self.age < 25:
            print("I am at college")
        else:
            print("I work as a Python developer!")
```
