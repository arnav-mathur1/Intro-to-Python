```python
class C2:
    """This is my second class!"""
    
    def person(self, x = 300):
        print("I'm " + str(x) + " years old")
        print()
        
    def outsider(self, name):
        print("{} is an outsider!".format(name))
        
    def insider(self):
        name = "Peter"
        print("{} is inside the class".format(name))
        
    def square(self, num):
        return num**2
    
    def decor1(self, func):
        print("Decorator 1 inside of a class!")
        
        def check(*args, **kwargs):
            f = func(*args, **kwargs)
            for i in range(f):
                print(i)
            statement = "The final value is " + str(f)
            return statement
        return check
```

```python
c = C2()
```

```python
c.square(3)
d = c.decor1(c.square)
```
