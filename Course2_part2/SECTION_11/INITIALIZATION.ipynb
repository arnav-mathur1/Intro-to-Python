```python
class C3:
    
    def talk(self, speak, code):
        self.speak = speak
        self.code = code
        self.action = "algorithm"
        
    def talking(self):
        print("I'd like to talk about " + self.speak)
        
    def programmer(self):
        print("My code is " + self.code)
        
    def running(self):
        print(f"I ran my {self.speak} {self.action}")
```

```python
d = C3()
```

```python
d.talk("javascript", "slow")
d.talking()
d.programmer()
d.running()
```

```python
d.action = "decorator"
```

```python
d.running()
```

```python
class Animals:
    
    """This is a class of animals"""
    a = 100
    b = 245
    c = 12000
    
    def __init__(self, species, action, cute, talk):
        self.species = species
        self.action = action
        self.cute = cute
        self.talk = talk
        
    def speak(self):
        print("I can {}".format(self.talk))
        
    def animal(self):
        print("I'm a {} and I can {}".format(self.species, self.action))
        
    def eating(self):
        print(f"{self.species} is super hungry!")
```

```python
dog = Animals("husky puppy", "run", "2 billion", "Hoooowl")
parrot = Animals("lovbird", "fly", "23 trillion", "talk to humans")
piggy = Animals("piglet", "run like a dog", "3.4 gazillion", "oink")
```

```python
dog.speak()
dog.animal()
dog.eating()
```

```python
dog.species = "jack russell"
dog.talk = "woof!"
dog.animal()
```
