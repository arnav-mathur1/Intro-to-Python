```python
class Calculator:
    
    def __init__(self, x):
        self.x = x
        
    def __add__(self, val):
        return self.x + val.x
    
    def __sub__(self, val):
        return self.x - val.x
    
    def __mul__(self, val):
        return self.x * val.x
    
    def __truediv__(self, val):
        return self.x / val.x
    
    def __floordiv__(self, val):
        return self.x // val.x
    
a = Calculator(34)
b = Calculator(12)
```

```python
print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
```

```python
class Animal():
    
    def __init__(self, species, age):
        self.species = species
        self.age = age
        self.index = 0
        self.pets = ["gracie", "tommy", "gizmo", "enzo", "gilbert", "mary"]
        
    def __str__(self):
        return "I'm a super cute {}".format(self.species)
    
    def __len__(self):
        return len(self.pets)
    
    def __getitem__(self, num):
        return self.pets[num]
    
    def __del__(self):
        self
        
    def __next__(self):
        word = self.pets[self.index]
        self.index += 1
        return word

    def __delitem__(self, num):
        del(self.pets[num])
        
    def __getslice__(self, num1, num2):
        return self.pets[num1: num2]
    
    def __call__(self):
        for i in self.pets:
            print(i)
            
cat = Animal("kitten", 2)
dog = Animal("puppy", 3)
```

```python
print(cat)
print(dog)
```
