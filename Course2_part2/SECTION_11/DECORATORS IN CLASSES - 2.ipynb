```python
class Delivery:
    service_charge = 1.15  
    extras = "sweetcorn soup"
#     prices = [1.0, 0.85, 0.7, 0.50]
    
    def __init__(self):
        
        self.menu = {1: ["coconut curry", 7.50], 
                     2: ["spicy tofu",  5.25], 
                     3: ["veg noodles", 8.75]} 
        self.order = []
        
    def Choose_in_menu(self):
        for i in range(1, len(self.menu) + 1):
            print(i, self.menu[i][0] + " $" + str(self.menu[i][1]))
            
        print("Please enter the dish numbers you'd like")
        
        while True:
            self.food = input("")
            try:
                num = abs(int(eval(self.food)))
                
            except:
                print("Please enter a valid number")
                continue
                
            if self.food == "" or num == 0 and num < 4:
                print("Thank you. Your order is complete")
                break               
            elif num >= 4:
                print("Please enter a number between 1 and 3")
            else:
                self.order.append(self.menu[num][1])
                        
    def Pay(self):
        total = sum(self.order) * Delivery.service_charge
        print("That will be total of $" + str(round(total, 2)))
        if total > 30:
            print(f"We've also included a free {Delivery.extras}")           
        else:
            None
            
    def Decision_Time(self, func):
        print("How long it takes for the customer to choose?")
        def Customer_Time(*args, **kwargs):
            import time
            start = time.time()
            dec1 = func()
            end = time.time()
            
            duration = round(((end-start)*1000)/1000,2)
            print("The time was {} seconds".format(duration))
            
            return dec1
        return Customer_Time
    
    def Type_Details(self, func):
        print("Entering customer details before payment")
        
        def Customer_Details(*args, **kwargs):      
            import datetime as dt
            dec2 = func
            date = dt.datetime.now().strftime("%Y/%m/%d")
            while True:
                first = input("Your first name: ").lower()
                last = input("Your last name: ").lower()
                address = input("Your city: ").upper().replace(" ", "-")

                customer = [first, last, address]
                print("-"*20)
                print("Thank you. Your order will be delivered in 30 minutes!")

                path = "/Users/arnavmathur/Desktop/CUSTOMERS/" 
                with open(path + "Customer_Data.txt", "r") as f:
                    data = f.readlines()

                    check = " ".join(data).replace("\n", "").split()

                    case = [i for i in check if "/" not in i]

                    loc = [case[i] for i in range(len(case)) if case[i] == case[i].upper()]

                    names = [case[i] for i in range(len(case)) if case[i] == case[i].lower()]

                    nested_names = [names[i:i+2] for i in range(len(names)) if i % 2 == 0]

                    fin = [nested_names[i] + [loc[i]] for i in range(len(nested_names))]

                if customer in fin:
                    print("Regular customer")
                    dec2()
                    break
                else:
                    print("Welcome new customer!")
                    print("You have 20% off your first order!")
                    Delivery.service_charge = 0.95
                    dec2()
                    Delivery.service_charge = 1.15
                    with open(path + "Customer_Data.txt", "a") as f:
                        f.write("\n")
                        f.write("\n" + str(date))
                        f.write("\n" + customer[0])
                        f.write("\n" + customer[1])
                        f.write("\n" + customer[2])
                        f.write("\n")
                        break
                        
        return Customer_Details 
```
