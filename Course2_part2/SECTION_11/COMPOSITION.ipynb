```python
class Basic():
    x = 20
    y = 8
    def __init__(self):
        pass
        
    def Addition(self):
        return Basic.x + Basic.y
    
    def Subtraction(self):
        return Basic.x - Basic.y
    
    def Times_table(self):
        for i in range(1, Basic.y):
            print("%i * %i = %i" % (i, Basic.y, i*Basic.y))
            
b = Basic()
b.Addition()
b.Subtraction()
b.Times_table()
```

```python
class Advanced(Basic):    
    def __init__(self):
        pass 
    
    def Multiply(self):
        return self.x * self.y

a = Advanced()
a.Times_table()
a.Multiply()
```

```python
class Comp():
    def __init__(self):
        self.base = Basic()
        
    def Multiply(self):
        return self.base.x * self.base.y
    
    def Times_y(self):
        return self.base.Times_table()
    
c = Comp()
c.Multiply()
c.Times_y()
```
