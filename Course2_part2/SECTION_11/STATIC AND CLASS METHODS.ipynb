```python
class Charity:
    
    funds = 40000
    donation = 0
    
    def __init__(self, donation):
        self.donation = donation
    
    def King(self, add):
        Charity.funds = Charity.funds + add
        print("I have given a total of $" + str(Charity.funds))
        
    def Self_Reset(self):
        Charity.funds = 45000
        
    @classmethod
    def Cls_Reset(cls):
        cls.funds = 50000
        
    @staticmethod
    def Static_Reset():
        Charity.funds = 55000
        
    @classmethod
    def Total_funds(cls):
        cls.funds = cls.funds + cls.donation
        print("Total money in our treasury is ${}".format(cls.funds))
        
    @classmethod
    def From_donor(cls, donated):
        cls.donation = eval(donated.split("$")[-1])
        
    @staticmethod
    def Allocates(cause1, cause2, percent):
        funds = Charity.funds
        total = 1.0
        c1 = total - percent
        c2 = percent
        print(f"The {cause1} charity will recieve ${funds*c1} and the {cause2} recieves {funds*c2}")
```
