```python
class Compare:    
    value = 250
    
    def __init__(self): 
        self.value = 300 
        
    def Self_Wallet(self):
        print("I have a total of ${}".format(self.value))
        
    @staticmethod
    def Static_Wallet():
        print("I have a total of ${}".format(Compare.value))
        
    @classmethod
    def Cls_Wallet(cls):
        print("I also have a total of ${}".format(cls.value))
        
c = Compare()
c.Self_Wallet()
c.Static_Wallet()
c.Cls_Wallet()
```

```python
class Course:
    sub = "Python"
    @classmethod
    def Subject(cls):
        print("The subject is on {} programming".format(cls.sub))

class Teacher:
    name = "Prof Xavier" 
    class_room = Course()
    
    @classmethod
    def Lecturer(cls):
        cls.class_room.Subject()
        print("My name is {} and I'll be teaching {}".format(cls.name, cls.class_room.sub))
    
class Student(Course):
    sub = "OOP in Python"
    
t = Teacher()
t.Lecturer()
print()
s = Student()
s.Subject()
```

```python
class cloud:
    green_materia = "Healing winds!"
    def __init__(self): 
        self.limit_break = "Omni Slash"
    
    def Final_fantasy(self):
        print(f"Limit breaks with {self.limit_break}")

class tifa(cloud): 
    red_materia = "Fire storm!!"
    def __init__(self):
        self.limit_break = "Final Heaven"
    
    @classmethod
    def Magic1(cls):
        print("Spell 1: {}".format(cls.green_materia))

class barret(tifa):
    def __init__(self):
        self.limit_break = "Catastrophe"
    
    @classmethod
    def Magic2(cls): 
        print("Spell 2: {}".format(cls.red_materia))

class sephiroth(barret):
    def __init__(self): 
        self.limit_break = "Super Nova" 
        self.wings = "flight!!!!"
    
    def Magic3(self):
        super(sephiroth, self).Magic1() 
        super(sephiroth, self).Magic2()
        print("Spell 3: {}".format(self.wings))

s = sephiroth()
s.Final_fantasy()
print()
s.Magic3()
```

```python
class Inventory:
    def __init__(self):
        self.items = ["sword", "cross bow", "axe", "shield", "staff"] 
        self.index = 0
    @classmethod
    def __str__(cls):
        return f"I'm a {cls.name} of the castle!"
    
    def __len__(self):
        return len(self.items)
    
    def __next__(self):
        check = self.items[self.index] 
        self.index += 1
        return check

class Knight(Inventory): 
    name = "Lancelot"
    @staticmethod
    def __call__():
        print("I shall use my sword!") 

class Wizard(Inventory):
    name = "Merlin"
    @staticmethod
    def __call__():
        print("I shall use my staff!")

k = Knight() 
w = Wizard() 
k()
w()
print()
print(k)
print(w)
```
