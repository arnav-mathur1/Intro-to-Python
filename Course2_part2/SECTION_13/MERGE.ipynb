```python
import pandas as pd
```

```python
data1 = {"runaways": ["alex", "nico", "chase"], "brain_power": [10, 8, 7]}
df1 = pd.DataFrame(data1)

data2 = {"runaways": ["alex", "nico", "chase"], "power_level": [20, 5, 12]}
df2 = pd.DataFrame(data2)
```

```python
df1
```

```python
pd.merge(df1, df2, on = "runaways", how = "outer", indicator = True)
```

```python
df1.loc[3] = "gert"
df1["brain_power"].loc[3] = 3
df1

df2.loc[3] = "karolina"
df2["power_level"].loc[3] = 26
df2
```

```python
df1["health_bar"] = 99, 100, 76, 54
df2["health_bar"] = 99, 100, 76, 2000
df2
```

```python
pd.merge(df1, df2, on = ["runaways", "health_bar"], how = "outer")
```
