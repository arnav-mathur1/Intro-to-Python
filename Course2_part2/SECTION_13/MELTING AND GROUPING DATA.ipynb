```python
import pandas as pd
```

```python
food = {"Dish": ["Curry", "Noodles", "Pasta", "Salad", "Pizza"],
        "GER": [10, 5, 4.5, 2.5, 9],
        "USA": [11, 5, 7, 3, 11],
        "UK": [12, 6, 4, 2.5, 8]}
```

```python
df_food = pd.DataFrame(food)
df_food
```

```python
df_food.shape
```

```python
df = pd.melt(df_food, id_vars = "Dish", var_name = "Country", value_name = "Price")
```

```python
pt = df.pivot_table(index = "Country", values = "Price", columns = "Dish", aggfunc = sum)
pt.loc["USA"]
```

```python
dish = df.set_index("Country")
dish.shape
d = dish.groupby("Country")
d
```

```python
d.get_group("USA")
```

```python
for i in d:
    print(i)
```

```python
gen = iter(d)
gen
```
