```python
import random as rd, datetime as dt
print("WELCOME PLAYER!")
print(" ")
game = input("TYPE START FOR NEW SCOREBOARD, OR CONTINUE TO UPDATE SCOREBOARD ").lower()

def Clear_Game():
    game = open("Game.txt", "w")
    game.write("SCORE BOARD                 |       DATE AND TIME")
    game.close()

def Results(p1_score, p2_score): 
    print("GAME OVER!")
    final = ("Player1 = {} VS Player 2 = {}".format(p1_score, p2_score))
    print(final)
    date = dt.datetime.now().strftime("      %m/%d/%Y  %H:%M:%S")
    return final, date

def Timer(score):
    f = open("Game.txt", "a")
    f.write("\n" + score[0])
    f.write(score[1])

if game == "start":
    print("STARTING!")
    print(" ")
    Clear_Game()
    
    play = 3
    sign = ["scissors", "paper", "rock"]
    p1_score, p2_score = 0, 0
    while play > 0:

        player1 = input("P1 sign: ").lower()

        if player1 in sign and player1 != "quit":

            player2 = rd.choices(sign, [1/3]*3)[0]
            print("P2 sign: ", player2)

            if player1 == sign[0] and player2 == sign[1] or\
            player1 == sign[1] and player2 == sign[2] or\
            player1 == sign[2] and player2 == sign [0]:
                print("Player1 wins!")
                print(" ")
                p1_score += 1

            elif player2 == sign[0] and player1 == sign[1] or\
            player2 == sign[1] and player1 == sign[2] or\
            player2 == sign[2] and player1 == sign [0]:
                print("Player2 wins!")
                print(" ")
                p2_score += 1

            else:
                print("DRAW!!!")
                print(" ")

            play -= 1

        elif player1 == "quit":
                res = Results(p1_score, p2_score)
                Timer(res)
                break


        elif player1 not in sign and player1 != "quit":
            print("Please enter a valid sign")
            print(" ")

    if play == 0:
        res = Results(p1_score, p2_score)
        Timer(res) 
        
elif game != "start" and game != "continue":
    print("PLEASE ENTER START OR CONTINUE NEXT TIME")
    print(" ")
    input("PRESS ENTER TO EXIT")

else:
    print("CONTINUED!")
    print(" ")
    
    play = 3
    sign = ["scissors", "paper", "rock"]
    p1_score, p2_score = 0, 0
    while play > 0:

        player1 = input("P1 sign: ").lower()

        if player1 in sign and player1 != "quit":

            player2 = rd.choices(sign, [1/3]*3)[0]
            print("P2 sign: ", player2)

            if player1 == sign[0] and player2 == sign[1] or\
            player1 == sign[1] and player2 == sign[2] or\
            player1 == sign[2] and player2 == sign [0]:
                print("Player1 wins!")
                print(" ")
                p1_score += 1

            elif player2 == sign[0] and player1 == sign[1] or\
            player2 == sign[1] and player1 == sign[2] or\
            player2 == sign[2] and player1 == sign [0]:
                print("Player2 wins!")
                print(" ")
                p2_score += 1

            else:
                print("DRAW!!!")
                print(" ")

            play -= 1

        elif player1 == "quit":
                res = Results(p1_score, p2_score)
                Timer(res)
                break


        elif player1 not in sign and player1 != "quit":
            print("Please enter a valid sign")
            print(" ")

    if play == 0:
        res = Results(p1_score, p2_score)
        Timer(res) 
```
