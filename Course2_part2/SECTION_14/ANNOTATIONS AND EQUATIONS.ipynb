```python
import matplotlib.pyplot as plt, numpy as np
%matplotlib inline
```

```python
plt.figure(figsize = (6, 4))

alpha = np.random.randint(1, 50, 10)
beta = np.random.randint(1, 50, 10)

sum_x = sum(alpha); sum_y = sum(beta)
if sum_x > sum_y:
    val = ">"
    num = sum_x
    let = "alpha"
else:
    val = "<"
    num = sum_y
    let = "beta"

plt.plot(alpha, lw = 5); plt.plot(beta, lw = 5)
plt.xlabel("X"); plt.ylabel("Y")
plt.text(10, 12, f"""$sum_\{let}^\infty \{let}_i = {num}$""", fontsize = 40)

plt.legend(["alpha", "beta"], fontsize = 25, loc = (1.1, 0.5))

plt.title(r"""EQUATIONS:
$\alpha {} \beta$""".format(val), fontsize = 30)
plt.show()
```

```python
import time
t = np.arange(0, 5, 0.001)
s = np.cos(2 * np.pi * t)

for i in range(6):
    plt.plot(t, s, lw = 2, color = "b")
    
    plt.annotate("local max {}".format(i), xy = (i, 1), xytext = (3, 1.5), fontsize = 15,
                 arrowprops = dict(facecolor = "r", shrink = 2))
    
    plt.annotate("local min {}".format(i), xy = (0.5 + i, -1), xytext = (3, -1.5), fontsize = 15,
                 arrowprops = dict(facecolor = "green", shrink = 2))
    time.sleep(1)
    plt.ylim(-2, 2)
    plt.show()
```
