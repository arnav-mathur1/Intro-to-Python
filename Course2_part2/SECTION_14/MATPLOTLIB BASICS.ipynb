```python
import matplotlib.pyplot as plt, numpy as np
%matplotlib inline
```

```python
x = np.arange(1, 10)**2
y = np.arange(1, 20, 2)**2
z = np.arange(1, 40, 4)**2

plt.plot(x)
plt.show()
```

```python
plt.figure(figsize = (10, 4))
plt.plot(x, label = "X", lw = 5, marker = "^", markersize = 15)
plt.plot(y, label = "Y", lw = 5, marker = "s", markersize = 15)
plt.plot(z, label = "Z", lw = 5, marker = "o", markersize = 15)
plt.xlabel("TIME", fontsize = 15)
plt.ylabel("PRICE", fontsize = 15)
plt.title("GRAPH_1", fontsize = 21, color = "darkred", loc = "center")
plt.xticks(fontsize = 15); plt.yticks(fontsize = 15)
plt.legend(fontsize = 18, loc = (1.1, 0.3))
plt.savefig(path + "My_first_graph.png", bbox_inches = "tight", dpi = 100)
plt.show()
```

```python
data = [np.random.randint(1, 20, 10), np.random.randint(1, 20, 10), np.random.randint(1, 20, 10)]
data

for i in range(len(data)):
    plt.plot(data[i], label = ["A", "B", "C"][i], marker = ["o", "d", "s"][i],
             markersize = 15, color = ["pink", "b", "g"][i], lw = 5,
             markerfacecolor = "r", linestyle = ["--", "-", "-."][i], markeredgewidth = 5)
    plt.xticks(fontsize = 18); plt.yticks(fontsize = 18)
    plt.grid(True)
    plt.legend(["A", "B", "C"][i], fontsize = 20, loc = (1.1, 0.3))
    plt.show()
```
