```python
import matplotlib.pyplot as plt, numpy as np
%matplotlib inline
```

```python
np.random.seed(12)
num = np.random.randn(100000)
plt.hist(num, bins = 60, color = "b", orientation = "vertical", histtype = "bar", 
         width = 0.2)
plt.show()
```

```python
np.random.seed(10)
x = np.random.normal(1, 2, 5000)
y = np.random.normal(-1, 3, 5000)
z = np.random.normal(-2, 4, 5000)
```

```python
data = np.vstack([x, y, z]).T
data
bins = np.linspace(-10, 10, 30)
```

```python
plt.figure(figsize = (10, 6))
plt.hist(data, bins, label = ["x", "y", "z"], stacked = True, histtype = "step",
         orientation = "horizontal")
plt.legend(loc = "upper right", fontsize = 15)
plt.show()
```

```python
from matplotlib.colors import LogNorm as lg
```

```python
np.random.seed(20)
x = np.random.randn(100000)
y = np.random.randn(100000) + 5

plt.hist2d(x, y, bins = 200, norm = lg())
plt.colorbar(orientation = "vertical")
plt.show()
```
