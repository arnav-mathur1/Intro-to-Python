```python
import matplotlib.pyplot as plt, numpy as np
%matplotlib inline
```

```python
for page in range(2):
    fig = plt.figure()
    plt.plot(np.random.randint(1, 50, 10), color = ["r", "b"][page], lw = 5)
    
    plt.savefig(f"/Users/arnavmathur/Desktop/MATPLOTLIB/GRAPH_{page}.pdf",
                bbox_inches = "tight", dpi = 300)
```

```python
from matplotlib.backends.backend_pdf import PdfPages as pdf

with pdf("/Users/arnavmathur/Desktop/MPL_Graphs.pdf") as image:
    for i in range(3):
        fig = plt.figure()
        plt.plot(np.random.randint(1, 50, 12), color = ["b", "orange", "m"][i], lw = 7)
        plt.title("Graph_{}".format(i))
        
        image.savefig(fig, bbox = "tight", dpi = 300)
        plt.show()
```
