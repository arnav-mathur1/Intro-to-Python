```python
import matplotlib.pyplot as plt, numpy as np
%matplotlib inline
```

```python
x = np.random.randint(1, 100, 20)
y = np.random.randint(1, 100, 20)
y2 = np.random.randint(1, 100, 20)
x2 = np.random.randint(1, 100, 20)
plt.figure(figsize = (6, 4))

plt.scatter(x, y, label = "first", color = "b", s = 250, marker = "d",
            edgecolors = "r", linewidths = 3)
plt.scatter(x2, y2, label = "second", color = "orange", s = 250, marker = "h",
            edgecolors = "r", linewidths = 3)
plt.xlabel("X", fontsize  = 15); plt.ylabel("Y", fontsize = 15)

plt.xticks(size = 12); plt.yticks(size = 12)
plt.title("Cool Shapes Plot! \n Udemy 2020", fontsize = 18)
plt.legend(fontsize = 12, loc = (1.1, 0.6))
plt.show()
```

```python
x = np.random.randint(1, 100, 20)
y = np.random.randint(1, 100, 20)
y2 = np.random.randint(1, 100, 20)
x2 = np.random.randint(1, 100, 20)
plt.figure(figsize = (6, 4))

area = np.random.randint(1, 100, 20)**2
color = np.random.randint(1, 100, 20)

plt.scatter(x, y, s = area, c = color, marker = "o", edgecolors = "k", linewidths = 2)

plt.xticks(size = 12); plt.yticks(size = 12)
plt.title("Cool Bubbles! \n Udemy 2020", fontsize = 18)
plt.show()
```

```python
x = np.random.randint(1, 100, 100)
y = np.random.randint(1, 100, 100)

plt.figure(figsize = (6, 4))

for a, b in zip(x, y):
    if a >= 50 and b >= 50:
        plt.scatter(a, b, s = 400, c = "red", marker = "s")
    
    else:
        plt.scatter(a, b, s = 400, c = "g", marker = "^")
        
plt.show()
```
