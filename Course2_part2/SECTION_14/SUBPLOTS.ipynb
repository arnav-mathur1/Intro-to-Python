```python
import matplotlib.pyplot as plt, numpy as np
%matplotlib inline
```

```python
plt.subplot2grid((4, 4), (0, 0))
plt.subplot2grid((4, 4), (3, 3))
plt.subplot2grid((4, 4), (3, 0))
plt.subplot2grid((4, 4), (0, 1), colspan = 3)
plt.subplot2grid((4, 4), (1, 0), rowspan = 2)
plt.subplot2grid((4, 4), (2, 1), rowspan = 2, colspan = 2)
plt.subplot2grid((4, 4), (1, 3), rowspan = 2)
plt.subplot2grid((4, 4), (1, 1), colspan = 2)

plt.tight_layout()
plt.show()
```

```python
x = np.random.randint(1, 50, 20)
y = np.random.randint(1, 50, 20)
z1 = np.random.randint(1, 50, 40)
z2 = np.random.randint(1, 50, 40)

plt.figure(figsize = (8, 6))

plt.subplot2grid((3, 3), (0, 1))
plt.plot(x)
plt.title("A")

plt.subplot2grid((3, 3), (0, 2))
plt.plot(y, c = "r", marker = "s")
plt.title("B")

plt.subplot2grid((3, 3), (1, 1), colspan = 2)
plt.scatter(z1, z2, marker = "^", c = "g", s = 100)
plt.title("C")
plt.grid(True)

plt.subplot2grid((3, 3), (0, 0), rowspan = 2)
plt.imshow(np.random.randint(1, 100, 40).reshape(4, 10).T, cmap = "rainbow")
plt.title("D")
plt.colorbar()

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize = (8, 6))

plt.subplot(122)
plt.title("bar horizontal")

plt.barh([0, 1, 2, 3], [10, 20, 33, 12], height = 0.65, color = "c")

plt.subplot(241)
plt.title("bar")
plt.bar([0, 1, 2], [120, 140, 105], width = 0.7, color = "m")

plt.subplot(245)
plt.title("line plot")
plt.plot(np.random.randint(1, 40, 10), marker = "^", c = "g")

plt.subplot(142)
plt.title("heat map")
plt.imshow(np.random.randint(1, 50, 20).reshape(10, 2), cmap = "rainbow")
plt.colorbar()

plt.tight_layout()
plt.show()
```
