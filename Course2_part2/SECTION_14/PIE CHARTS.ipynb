```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
data = [5, 25, 50, 20]
sum(data)
plt.pie(data)
plt.show()
```

```python
pets = "CATS", "DOGS", "HEDGEHOGS", "COWS"
colors = "m", "c", "r", "y"
sizes = 15, 30, 45, 10
explosion = 0.1, 0.0, 0.0, 0.2
```

```python
plt.figure(figsize = (8, 8))
plt.pie(sizes, explode = explosion, labels = pets, colors = colors, 
        autopct = "%1.1f%%", shadow = True, radius = 1, startangle = 0)
plt.legend(pets, fontsize = 20, loc = (1.1, 0.4))
plt.show()
```
