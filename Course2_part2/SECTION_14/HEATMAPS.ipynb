```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
import numpy as np, pandas as pd
```

```python
data = np.random.randint(1, 50, 40).reshape(4, 10).T

df = pd.DataFrame(data, columns = ["AAPL", "MSFT", "YHOO", "BAC"])
df
```

```python
plt.imshow(df, cmap = "rainbow") # terrain, jet, hsv, plasma
plt.colorbar()
plt.show()
```

```python
compare = df.T[(df.T > 10) & (df.T < 30)]

plt.imshow(compare, cmap = "rainbow")
plt.colorbar()
plt.show() # white spaces are NaN values
```

```python
compare
```

```python
df2 = df.corr()
df2
```

```python
plt.imshow(df2, cmap = "rainbow")
plt.colorbar()
plt.xticks([0, 1, 2, 3], list(df))
plt.yticks(list(range(4)), list(df))
plt.show()
```

```python
df3 = np.array(df2)
df3
```

```python
for x in range(df3.shape[1]):
    for y in range(df3.shape[0]):
        plt.text(x, y, "%.2f" % df3[x, y], horizontalalignment = "center",
                 verticalalignment = "center")

plt.title("HEAT MAP", fontsize = 20)
plt.imshow(df3, cmap = "rainbow")
plt.yticks([0, 1, 2, 3], list(df))
plt.xticks(list(range(4)), list(df))
plt.colorbar()
plt.show()
```
