```python
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
import numpy as np, random as rd, pandas as pd
```

```python
data = [10, 25, 100, 40]
register = 0, 1, 2, 3
pets = "CAT", "DOG", "HEDGEHOG", "BUNNY"
plt.figure(figsize = (8, 4))

# b = plt.bar(register, data, width = 0.8, color = ("m", "r", "b", "g"))
b = plt.barh(register, data, height = 0.8, color = ("m", "r", "b", "g"))

plt.title("FAVORITE PETS", fontsize = 20)

# plt.xticks(register, pets)
plt.yticks(register, pets)

plt.legend(b, pets, fontsize = 12)
plt.show()
```

```python
rd.seed(12)
drinks = [rd.sample((range(1, 20)), 5), rd.sample((range(1, 20)), 5), rd.sample((range(1, 20)), 5)]
drinks
```

```python
x = np.arange(5)
days = "Mon", "Tue", "Wed", "Thur", "Fri"
plt.barh(x + 0.00, drinks[0], color = "b", height = 0.25, label = "Latte")
plt.barh(x + 0.25, drinks[1], color = "r", height = 0.25, label = "Cappuccino")
plt.barh(x + 0.50, drinks[2], color = "g", height = 0.25, label = "Tea")
plt.title("HOT DRINKS", fontsize = 20)
plt.yticks(x, days)
plt.legend(fontsize = 15, loc = (1.1, 0.4))
plt.show()
```

```python
x = np.arange(5)
days = "Mon", "Tue", "Wed", "Thur", "Fri"
plt.barh(x, drinks[0], left = np.sum(drinks[:0], axis = 0), label = "Latte") # change left to bottom and barh to bar
plt.barh(x, drinks[1], left = np.sum(drinks[:1], axis = 0), label = "Cappuccino")
plt.barh(x, drinks[2], left = np.sum(drinks[:2], axis = 0), label = "Tea")
plt.title("HOT DRINKS", fontsize = 20)
plt.yticks(x, days)
plt.legend(fontsize = 15, loc = (1.1, 0.4))
plt.show()
```

```python
colors = "b", "orange", "g"
for i in range(3):
    plt.barh(x, drinks[i], left = np.sum(drinks[i], axis = 0), color = colors[i]) 
    plt.yticks(x, days)
    
plt.show()
```
