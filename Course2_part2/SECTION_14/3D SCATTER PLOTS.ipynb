```python
import matplotlib.pyplot as plt, numpy as np
from mpl_toolkits.mplot3d import Axes3D
%matplotlib notebook
```

```python
fig = plt.figure()
ax1 = fig.add_subplot(131, projection = "3d")
ax2 = fig.add_subplot(133, projection = "3d")
x1 = np.random.randint(1, 50, 20)
y1 = np.random.randint(1, 50, 20)
z1 = np.random.randint(1, 50, 20)
a1 = np.random.randint(1, 50, 20)
b1 = np.random.randint(1, 50, 20)
c1 = np.random.randint(1, 50, 20)

ax1.scatter(x1, y1, z1, color = "red", marker = "^", s = 250, zdir = "y")
ax1.scatter(a1, b1, c1, color = "blue", marker = "s", s = 250, zdir = "y")

ax1.legend(["square", "triangle"], fontsize = 14, loc = (1.2, 0.2))

x2 = np.random.randint(1, 50, 20)
y2 = np.random.randint(1, 50, 20)
z2 = np.random.randint(1, 50, 20)
a2 = np.random.randint(1, 50, 20)
b2 = np.random.randint(1, 50, 20)
c2 = np.random.randint(1, 50, 20)

ax2.scatter(x2, y2, z2, color = "orange", marker = "o", s = 250, zdir = "x")
ax2.scatter(a2, b2, c2, color = "m", marker = "d", s = 250, zdir = "x")

ax2.legend(["circle", "diamond"], fontsize = 14, loc = (-1.2, 0.5))

plt.show()
```
