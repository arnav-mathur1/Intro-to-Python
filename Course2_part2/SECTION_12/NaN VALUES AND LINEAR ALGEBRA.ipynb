```python
import numpy as np
```

```python
arr = np.arange(1, 21, dtype = "float64").reshape(2, 10).T
arr
```

```python
# NaN = Not a Number
arr[5] = np.nan, np.nan
arr[2] = 3.0, np.nan
arr[7:9, 0:1] = np.nan
arr
```

```python
np.nan_to_num(arr) # CONVERTS NaN VALUES TO 0
np.nan_to_num(arr).min()
```

```python
np.nanmin(arr)
```

```python
arr2 = np.array([10, 20, np.nan, 30, np.nan])
arr2[~np.isnan(arr2)]
```
