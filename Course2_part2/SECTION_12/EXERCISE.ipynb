```python
class Compare1: 
    @classmethod
    def Loop(cls, num= 10000):
        val = list(range(1, num, 1))
        
        numbers = []        
        for i in val:
            res = i*25
            numbers.append(res) 
        return numbers

c2 = Compare1()
c2.Loop(10)
```

```python
class Compare2:
    import numpy as np
    
    @classmethod
    def Loop(cls, num = 10000):
        val = list(range(1, num, 1)) 
        numbers = cls.np.array(val)*25 
        return numbers.tolist()

c2 = Compare2()
c2.Loop(10)
```

```python
class Gen_numpy:
    import numpy as np
    
    @classmethod   
    def Array_gen(cls, *args): 
        cls.np.random.seed(12)
        case = []
        for i in args:
            res = i 
            case.append(res) 
            tup = tuple(case)
        arr1 = cls.np.random.randint(1, 100, size = tup) 
        
        gen = (i for i in arr1)
        return gen

gp = Gen_numpy()
g = gp.Array_gen(4, 3, 2, 3)
next(g)
```
