```python
def test(func):
   
    print("time decorator")
    
    def wrapper(*args, **kwargs):
        import time
        start = time.time() 
        f = func(*args, **kwargs)
        end = time.time()
        fin = end - start
        final = func.__name__ + " Time/milliseconds " + str(fin*1000)
        print(final)
        return f
    return wrapper
```

```python
check_time = test(add_up)
```

```python
def rey(func):
    print("decorator 1")
    
    def inner1(*args, **kwargs):
        f = func(*args, **kwargs)
        import datetime as dt
        date = dt.datetime.now().strftime("%m/%d/%Y %H:%M:%S")
        print("The date and time is", date)
        return f
    return inner1

def ben(func):
    print("decorator 2")
    
    def inner2(*args, **kwargs):
        f = func(*args, **kwargs)
        name = f.split()[5]
        agents = ["poe", "finn", "porg", "rey", "ben", "luke"]
        if name not in agents:
            print("First Order DETECTED!")
        else:
            print("You're part of the resistance, {}".format(name))
        return f
    return inner2
```
