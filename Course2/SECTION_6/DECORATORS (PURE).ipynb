```python
def study_time(func):
    print("Decorator 1")
    def first_inner1():
        print("Called before the function")
        func()
        print("Called after the function")
    return first_inner1
```

```python
study = study_time(greet)
```

```python
study()
```

```python
def Rey():
    for i in range(7, 10):
        print("Episodes {}".format(i))
# Rey()
```

```python
def solo(func):
    print("Decorator 1")
    def inner1(*args, **kwargs):
        print("*" * 30)
        func(*args, **kwargs)
        print("*" * 30)
        print("inner 1")
    return inner1

def kylo(func):
    print("Decorator 2")
    def inner2(*args, **kwargs):
        print("%" * 30)
        func(*args, **kwargs)
        print("--" * 15)
        print("inner 2")
    return inner2

def luke(func):
    print("Decorator 3")
    def inner3(*args, **kwargs):
        print("||" * 15)
        func(*args, **kwargs)
        print("<>" * 15)
        print("inner 3")
    return inner3
```

```python
starwars = luke(kylo(solo(Rey)))
```

```python
starwars()
```

```python
@luke
@kylo
@solo
def Rey():
    for i in range(7, 10):
        print("Episodes {}".format(i))
# Rey()
```
