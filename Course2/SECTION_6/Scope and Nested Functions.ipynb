```python
x = "cool"
def first_function():
    x = 45.123
    
    def second_function():
#         nonlocal x
        x = 33
        print("inner", x)
        
        def third_function():
            nonlocal x
            x = 1000
            print("extra inner", x)
            
            def fourth():
                global x
                x = "star wars"
                print("super inner = ", x)
            fourth()
        
        third_function()
    second_function()
    print("outer", x)

first_function()
print("global", x)
```
