```python
def Add(a, b, c, z=10):
    return a + b + c + z
Add(0, 0, 0, 0)
```

```python
def Tot(*args, z=10):
    
    s = 0
    for i in args:
        s += i
    print(args)
    return s + z
Tot(20, 30, 40, 100, 200)
```

```python
def Hotel(*args, price = 50):
    
    booked = []
    for i in args:    
        
        booked.append(i)
    print(f"Price per room ${price}")
    print(f"\nTotal for {len(booked)} people is ${price*len(booked)}")
    return booked
Hotel("Michael", "Gracie", "Akira", "Miki", "Alita")
```

```python
def Inventory(*args, **kwargs):
    
    players = []
    player_rpg = {}
    for i in args:
        players.append(i)
        player_rpg.update({i: kwargs})
        
    return player_rpg
rpg = Inventory("Michael", "Cloud", "Barret", a = "herb", b = "shield", c = "sword")
```
