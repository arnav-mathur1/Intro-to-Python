```python
def compute(*args):    
    import random as rd    
    case = []    
    for i in args:        
        case.append(i**2)    
    rd.seed(12)    
    sample = rd.sample((range(1, 100, 2)), k = 10)    
    return sum(case + sample)
```

```python
compute(10, 20, 30, 100) + compute(40, 50, 60, 70, 80, 90, 100) 
```

```python
def lang(prog1 = "PYTHON", prog2 = "DECORATOR"):    
    
    print("My favourite language is {}.".format(prog1))
    
def decor1(func):   
    print("Decorator 1")    
    def udemy(prog1, prog2):        
        
        print("The coolest part about {} is functions".format(prog1)) 
        f = func(prog1)
        print("...especially {} functions!".format(prog2))
        
        return f 
    return udemy
```

```python
u = decor1(lang)
```

```python
# @decor2
def comp(num):    
    case = []    
    for i in range(num):        
        res = i**2        
        case.append(res)    
    return case
```

```python
def decor2(func):
    print("decorator 2")
    def wrapper(*args, **kwargs):
        
        f = func(*args, **kwargs)
        rad = sum([i + 100 for i in f])
        
        return rad
    return wrapper
```

```python
c = (decor2(comp))
```

```python
c(10)
```

```python
def greet():    
    name = input("Enter your name: ")    
    print("Hello there, " + name,"!") 
    
def decor3(func):    
    print("Decorator 3")    
    def repeat(num, *args, **kwargs):        
        for i in range(num):            
            f = func(*args, **kwargs)         
            print("ID: 00{}".format(i+7))        
        
        return f   
    return repeat
```

```python
r = decor3(greet)
```

```python
r(3)
```
