```python
groceries = ["apples", "chips", "bread", "ice cream"]
prices = [2, 3, 1.2, 4.25]

for food in range(4):
    print(groceries[food], "=", prices[food])
```

```python
zip_shop = list(zip(groceries, prices))
list(zip_shop)

for g, p in zip_shop:
    print(g, "=", p)
```

```python
num1 = [100, 2, 90, 10]
num2 = [12, 7, 90, 50]
zip_num = zip(num1, num2)

for i, j in zip_num:
    if i > j:
        print(i)
    elif i < j:
        print(j)
    else:
        print(i, j)
```

```python
for i in range(4):
    if num1[i] > num2[i]:
        print(num1[i])
    elif num1[i] < num2[i]:
        print(num2[i])
    else:
        print(num1[i], num2[i])
```
