```python
nest1 = [[10, 20, 30], [3.5, 4.5, 5.5], ["sword", "hammer", "shield"]]

for i in range(3):
    print(nest1[i])
```

```python
weapons = []
for i in range(3):
    s = nest1[2][i].capitalize() + ": Power Level =" + str(nest1[0][i] + nest1[1][i])
    weapons.append(s)
weapons[2]
```

```python
for i in range(1, 8):
    print("{} * {} = {}".format(i, i, (i*i)))
```

```python
for k in range(1, 8):
    print("%f * %d = %d" % (k, k, (k*k)))
```

```python
tup1 = ((10, 20, 30), ("a", "b", "c"), (200, 400, 800))
# TUPLE UNPACKING!
for i in range(3):
    for x, y, z in tup1:
        print(x, y, z)
```

```python
cakes = ["chocolate", "lemon", "carrot", "vanilla"]

d1 = {"k1": tuple(cakes), "k2": [10, 20, 30, 40]}
d1["k2"]
```

```python
for d in d1["k2"]:
    print(d)
```

```python
for x in d1:
    for i in range(4):
        print(x, ":", d1[x][i])
```
