```python
stock = [] 
for i in range(10):    
    if i <= 5:       
        trade = "buy"    
    elif i > 5 and i <= 7:        
        trade = "hold"    
    else:        
        trade = "sell"    
    stock.append(trade.capitalize()) 
stock
```

```python
["Buy" if i <= 5
 else "Hold" if i > 5 and i <= 7
 else "Sell"
 for i in range(10)]
```

```python
pets = ["mouse", "cat", "dog", "pig", "cow", "lamb"] 
intelligence = [] 
for i in pets:    
    if "p" in i:        
        for k in pets:           
            if "d" in k:                
                res = i +"s are just as clever as {}s".format(k)                
                intelligence.append(res) 
intelligence
```

```python
pets = ["mouse", "cat", "dog", "pig", "cow", "lamb"] 
[i + "s are just as clever as {}s".format(k) for i in pets if "p" in i for k in pets if "d" in k]
```

```python
store = {} 
prices = [90, 120, 40, 200, 82] 
toys = ["car", "jet", "pony", "robot", "house"] 
for s in range(5):    
    if prices[s] < 100:        
        store[toys[s].upper()] = "$" + str(prices[s])
store
```

```python
d1 = {toys[s].upper(): "$" + str(prices[s]) for s in range(5) if prices[s] < 100}
d1
```
