```python
numbers = [20, 10, 30, 40, 50, 90, 93, 83, 73]
calc = []
for x in numbers:
    if x >= 50:
        res = x + 1
    else:
        res = x + 5
    calc.append(res)
calc
```

```python
[(x + 1) if x >= 50 else (x + 5) for x in numbers]
```

```python
move = list(range(7))
action = []
for x in move:
    if x == 2:
        res = "Run"
    elif x == 3:
        res = "Jump"
    elif x == 5:
        res = "Duck"
    else:
        res = "Walk"
    action.append(res)
action
```

```python
["Run" if x == 2 
 else "Jump" if x == 3 
 else "Duck" if x == 5 
 else "Walk" 
 for x in move]
```

```python
bought = []; sold = []
for x in range(5):
    if x % 2 == 0:
        b = "Buy"
        bought.append(b)
    else:
        s = "Sell"
        sold.append(s)
bought + sold
```

```python
["buy" for b in range(5) if b % 2 == 0] + ["sell" for s in range(5) if s % 2 == 1]
```

```python
stock = list(range(50, 200, 20))
stock
```

```python
trader = []
for p in stock:
    if p >= 50 and p <= 90:
        tr = "buy"
    elif p > 150:
        tr = "sell"
    else:
        tr = "hold"
    trader.append(tr)
trader
```

```python
["buy" if p >= 50 and p <= 90
 else "sell" if p > 150
 else "hold"
 for p in stock]
```

```python
result = []
for x in range(5):
    if x % 2 == 0:
        for y in range(5):
            if y % 2 == 1:
                result.append((x, y))
result
```
