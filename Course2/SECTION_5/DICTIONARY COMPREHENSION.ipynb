```python
alphabet = ["a", "b", "c", "d", "e"]
ran = range(10, 15)
em = {}
for i in range(5):
    if ran[i]**2 < 150:
        val = ran[i]**2
        s = {alphabet[i].upper(): val}
        em.update(s)
em
```

```python
d3 = {alphabet[x].upper(): ran[x]**2 for x in range(5) if ran[x]**2 < 150}
d3
```

```python
d4 = {}
menu = ["noodles", "curry", "pizza", "pasta"]
for i in range(len(menu)):
    d = {menu[i]: "$" + str((range(5, 20))[i])}
    d4.update(d)
d4
```
