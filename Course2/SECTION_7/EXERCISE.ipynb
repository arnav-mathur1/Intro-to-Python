```python
def mul(num1= 5, num2= 20, num3= 100):        
    return num1 + num2 + num3

mul()
```

```python
bowl = ["cherries", "orange", "apple", "melon", "figs"]       
def salad(fruit):            
    if fruit in bowl:                
        print(fruit)           
    else:                
        print("not in bowl")

salad("apple")
```

```python
bowl = ["cherries", "orange", "apple", "melon", "figs"]       
g2 = lambda x : fruit if x in bowl else "not in bowl"
g2("apple")
```

```python
def inside(num):            
    if num in list(range(10)):                
        return num**2            
    else:                
        print("outside")

inside(4)
```

```python
g3 = lambda num : num**2 if num in list(range(10)) else "outside"
g3(4)
```

```python
def error():
    while True:
        try:        
            val = int(eval(input("Enter an integer: ")))        
            g = lambda x: x*3 if x > 10 else "too low"        

            if g(val) > 20:                        
                print("Lambda is", g(val))                        
                print("Input value is", val)                        
                break            

            else:                        
                print("No output")        

        except:                
            print("Try again please")                
            continue

        finally:
            print("End code")
            
error()
```
