```python
def add(arg1= 10, arg2 = 40, arg3 = 50):
    return arg1 + arg2 + arg3

add()
```

```python
g = lambda x = 10, y = 40, z = 50 : x+y+z
g() # can override here
```

```python
def check(num):
    return num % 2 == 0 or num > 8

check(8)
```

```python
g2 = lambda num : num % 2 == 0 or num > 8
g2(8)
```

```python
def compare(a, b):
    if a > 10:
        return a
    else:
        return b

compare(4, 2)
```

```python
g3 = lambda a, b : a if a > 10 else b
g3(4, 2)
```

```python
def size(x):
    if x > 100:
        return "big"
    else:
        return "small"
    
size(90)
```

```python
g4 = lambda x : "big" if x > 100 else "small"
g4(90)
```
